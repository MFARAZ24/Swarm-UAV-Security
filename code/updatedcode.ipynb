```python
import pandas as pd
import numpy as np

# ============================================================
# STEP 2: Create clean RL dataset + reward dataset
# ============================================================

# =========================
# Load updated balanced datasets
# =========================
rl_df = pd.read_csv("uav_rl_dataset-updated.csv")
det_df = pd.read_csv("uav_detection_dataset-updated.csv")

print("===== LOADED DATASETS =====")
print("RL shape:", rl_df.shape)
print("Detection shape:", det_df.shape)

# =========================
# Clean RL dataset
# =========================
numeric_rl = rl_df.select_dtypes(include=[np.number])

print("\n===== BEFORE CLEANING =====")
print("RL NaN values:")
print(rl_df.isna().sum())

print("\nRL Inf values:")
print(np.isinf(numeric_rl).sum())

# Replace possible Inf with NaN
rl_df = rl_df.replace([np.inf, -np.inf], np.nan)

# Drop any remaining bad rows
# Your checked output should have no NaN/Inf, so this should keep all 28800 rows.
rl_df_clean = rl_df.dropna().copy()

print("\n===== AFTER CLEANING =====")
print("Clean RL shape:", rl_df_clean.shape)

print("\nClean RL NaN values:")
print(rl_df_clean.isna().sum())

print("\nClean RL Inf values:")
print(np.isinf(rl_df_clean.select_dtypes(include=[np.number])).sum())

# =========================
# Save clean RL dataset
# =========================
rl_df_clean.to_csv("uav_rl_dataset_clean.csv", index=False)

print("\nSaved clean RL dataset:")
print("uav_rl_dataset_clean.csv")

# ============================================================
# STEP 3: Build reward dataset
# ============================================================

group_cols = ["envId", "formationType", "timeStep", "receiverId"]

# For RL training, use benign detection rows only
benign_det = det_df[det_df["actual_target_malicious"] == 0].copy()

print("\n===== BENIGN DETECTION ROWS =====")
print("Benign detection shape:", benign_det.shape)

# Merge theta into detection rows
det_with_theta = benign_det.merge(
    rl_df_clean[group_cols + ["theta"]],
    on=group_cols,
    how="inner"
)

print("\nDetection rows matched with RL theta:", det_with_theta.shape)

# Static theta exceedance, used as baseline/reward reference
det_with_theta["exceeded_theta"] = (
    det_with_theta["delta_d"] > det_with_theta["theta"]
).astype(int)

# Aggregate per receiver-time state
exceedance_df = (
    det_with_theta
    .groupby(group_cols)
    .agg(
        benign_exceedance_rate=("exceeded_theta", "mean"),
        mean_delta_d=("delta_d", "mean"),
        max_delta_d=("delta_d", "max"),
        num_links=("delta_d", "count")
    )
    .reset_index()
)

print("\nExceedance dataframe shape:", exceedance_df.shape)

# Merge reward info back into RL dataset
rl_reward_df = rl_df_clean.merge(
    exceedance_df,
    on=group_cols,
    how="left"
)

# Fill any missing values safely
rl_reward_df["benign_exceedance_rate"] = rl_reward_df["benign_exceedance_rate"].fillna(0.0)
rl_reward_df["mean_delta_d"] = rl_reward_df["mean_delta_d"].fillna(0.0)
rl_reward_df["max_delta_d"] = rl_reward_df["max_delta_d"].fillna(0.0)
rl_reward_df["num_links"] = rl_reward_df["num_links"].fillna(0)

print("\n===== RL REWARD DATASET =====")
print("RL reward shape:", rl_reward_df.shape)

print("\nColumns:")
print(rl_reward_df.columns.tolist())

print("\n===== BENIGN EXCEEDANCE SUMMARY =====")
print(rl_reward_df["benign_exceedance_rate"].describe())

print("\n===== EXCEEDANCE BY ENVIRONMENT =====")
print(
    rl_reward_df
    .groupby("envId")["benign_exceedance_rate"]
    .describe()
)

print("\n===== EXCEEDANCE BY FORMATION =====")
print(
    rl_reward_df
    .groupby("formationType")["benign_exceedance_rate"]
    .describe()
)

print("\n===== NUM LINKS SUMMARY =====")
print(rl_reward_df["num_links"].describe())

# =========================
# Save reward dataset
# =========================
rl_reward_df.to_csv("uav_rl_reward_dataset.csv", index=False)

print("\nSaved reward dataset:")
print("uav_rl_reward_dataset.csv")
```

```python
import pandas as pd
import numpy as np

# =========================
# Load balanced datasets
# =========================
rl_df = pd.read_csv("uav_rl_dataset_clean.csv")
det_df = pd.read_csv("uav_detection_dataset-updated.csv")

# =========================
# Reward parameters
# =========================
alpha_rx = 0.05
lambda1 = 1.0
lambda2 = 0.05
lambda3 = 0.0   # no threshold-change penalty for manual fixed-threshold test

Tmax = 50.0

thresholds_to_test = [2, 5, 8, 10, 12, 15, 20, 25, 30, 35, 40]

group_cols = ["envId", "formationType", "timeStep", "receiverId"]

# Keep benign rows only
benign_det = det_df[det_df["actual_target_malicious"] == 0].copy()

results = []

# =========================
# Test multiple fixed thresholds
# =========================
for threshold in thresholds_to_test:

    temp = benign_det.copy()
    temp["exceeded"] = (temp["delta_d"] > threshold).astype(int)

    exceedance_df = (
        temp
        .groupby(group_cols)
        .agg(
            benign_exceedance_rate=("exceeded", "mean"),
            mean_delta_d=("delta_d", "mean"),
            max_delta_d=("delta_d", "max"),
            num_links=("delta_d", "count")
        )
        .reset_index()
    )

    merged = rl_df[group_cols].merge(
        exceedance_df,
        on=group_cols,
        how="inner"
    )

    merged["reward"] = (
        -lambda1 * (merged["benign_exceedance_rate"] - alpha_rx) ** 2
        -lambda2 * (threshold / Tmax)
    )

    summary = (
        merged
        .groupby("envId")
        .agg(
            threshold=("benign_exceedance_rate", lambda x: threshold),
            mean_exceedance=("benign_exceedance_rate", "mean"),
            std_exceedance=("benign_exceedance_rate", "std"),
            mean_reward=("reward", "mean"),
            rows=("reward", "count")
        )
        .reset_index()
    )

    results.append(summary)

reward_results = pd.concat(results, ignore_index=True)

print("===== REWARD RESULTS BY ENVIRONMENT =====")
print(reward_results)

best_by_env = (
    reward_results
    .sort_values(["envId", "mean_reward"], ascending=[True, False])
    .groupby("envId")
    .head(1)
)

print("\n===== BEST THRESHOLD PER ENVIRONMENT =====")
print(best_by_env)

# =========================
# Best threshold per env + formation
# =========================
formation_results = []

for threshold in thresholds_to_test:

    temp = benign_det.copy()
    temp["exceeded"] = (temp["delta_d"] > threshold).astype(int)

    exceedance_df = (
        temp
        .groupby(group_cols)
        .agg(
            benign_exceedance_rate=("exceeded", "mean"),
            num_links=("delta_d", "count")
        )
        .reset_index()
    )

    merged = rl_df[group_cols].merge(
        exceedance_df,
        on=group_cols,
        how="inner"
    )

    merged["reward"] = (
        -lambda1 * (merged["benign_exceedance_rate"] - alpha_rx) ** 2
        -lambda2 * (threshold / Tmax)
    )

    form_summary = (
        merged
        .groupby(["envId", "formationType"])
        .agg(
            threshold=("benign_exceedance_rate", lambda x: threshold),
            mean_exceedance=("benign_exceedance_rate", "mean"),
            mean_reward=("reward", "mean"),
            rows=("reward", "count")
        )
        .reset_index()
    )

    formation_results.append(form_summary)

formation_reward_results = pd.concat(formation_results, ignore_index=True)

best_by_env_formation = (
    formation_reward_results
    .sort_values(["envId", "formationType", "mean_reward"], ascending=[True, True, False])
    .groupby(["envId", "formationType"])
    .head(1)
)

print("\n===== BEST THRESHOLD PER ENVIRONMENT + FORMATION =====")
print(best_by_env_formation)

# Save outputs
reward_results.to_csv("reward_threshold_test_results_balanced.csv", index=False)
best_by_env.to_csv("best_threshold_by_environment_balanced.csv", index=False)
formation_reward_results.to_csv("reward_threshold_test_by_env_formation_balanced.csv", index=False)
best_by_env_formation.to_csv("best_threshold_by_env_formation_balanced.csv", index=False)

print("\nSaved:")
print("reward_threshold_test_results_balanced.csv")
print("best_threshold_by_environment_balanced.csv")
print("reward_threshold_test_by_env_formation_balanced.csv")
print("best_threshold_by_env_formation_balanced.csv")
```

```python
import numpy as np
import pandas as pd
import gymnasium as gym
from gymnasium import spaces


class UAVThresholdEnv(gym.Env):
    """
    Custom Gymnasium environment for UAV dynamic Phase-1 threshold learning.

    Observation:
        Receiver-level RL features.

    Action:
        Continuous threshold adjustment delta_T.

    Reward:
        Penalizes difference between benign exceedance rate and target alpha_rx,
        penalizes large threshold,
        penalizes abrupt threshold changes.
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        rl_path="uav_rl_dataset_clean.csv",
        det_path="uav_detection_dataset-updated.csv",
        alpha_rx=0.05,
        lambda1=1.0,
        lambda2=0.05,
        lambda3=0.01,
        Tmin=1.0,
        Tmax=50.0,
        delta_limit=2.0,
        random_start=True,
        max_episode_steps=500,
    ):
        super().__init__()

        self.rl_df = pd.read_csv(rl_path)
        self.det_df = pd.read_csv(det_path)

        self.alpha_rx = alpha_rx
        self.lambda1 = lambda1
        self.lambda2 = lambda2
        self.lambda3 = lambda3

        self.Tmin = Tmin
        self.Tmax = Tmax
        self.delta_limit = delta_limit

        self.random_start = random_start
        self.max_episode_steps = max_episode_steps
        self.episode_step = 0

        self.group_cols = ["envId", "formationType", "timeStep", "receiverId"]

        self.feature_cols = [
            "sigma_rssi_bar",
            "sigma_gps_bar",
            "snr_bar",
            "plr_bar",
            "relative_speed_bar",
            "relative_height_bar",
            "trusted_count",
            "trust_bar",
            "theta",
        ]

        self.benign_det = self.det_df[
            self.det_df["actual_target_malicious"] == 0
        ].copy()

        self.delta_lookup = {
            key: group["delta_d"].values.astype(np.float32)
            for key, group in self.benign_det.groupby(self.group_cols)
        }

        self.observation_space = spaces.Box(
            low=-np.inf,
            high=np.inf,
            shape=(len(self.feature_cols),),
            dtype=np.float32,
        )

        self.action_space = spaces.Box(
            low=np.array([-self.delta_limit], dtype=np.float32),
            high=np.array([self.delta_limit], dtype=np.float32),
            dtype=np.float32,
        )

        self.current_index = 0
        self.start_index = 0
        self.current_threshold = None

    def _get_obs(self):
        row = self.rl_df.iloc[self.current_index]
        obs = row[self.feature_cols].values.astype(np.float32)

        # Replace static theta with current dynamic threshold
        obs[-1] = self.current_threshold

        return obs

    def _get_key(self):
        row = self.rl_df.iloc[self.current_index]
        return tuple(row[col] for col in self.group_cols)

    def _compute_exceedance_rate(self, threshold):
        key = self._get_key()
        delta_values = self.delta_lookup.get(key, None)

        if delta_values is None or len(delta_values) == 0:
            return 0.0

        return float(np.mean(delta_values > threshold))

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)

        self.episode_step = 0

        if self.random_start:
            max_start = max(0, len(self.rl_df) - self.max_episode_steps - 1)
            self.start_index = int(self.np_random.integers(0, max_start + 1))
        else:
            self.start_index = 0

        self.current_index = self.start_index
        self.current_threshold = float(self.rl_df.iloc[self.current_index]["theta"])

        obs = self._get_obs()
        row = self.rl_df.iloc[self.current_index]

        info = {
            "start_index": self.start_index,
            "index": self.current_index,
            "threshold": self.current_threshold,
            "envId": int(row["envId"]),
            "formationType": int(row["formationType"]),
            "timeStep": int(row["timeStep"]),
            "receiverId": int(row["receiverId"]),
        }

        return obs, info

    def step(self, action):
        delta_T = float(np.clip(action[0], -self.delta_limit, self.delta_limit))

        old_threshold = self.current_threshold

        self.current_threshold = float(
            np.clip(
                self.current_threshold + delta_T,
                self.Tmin,
                self.Tmax,
            )
        )

        benign_exceedance_rate = self._compute_exceedance_rate(
            self.current_threshold
        )

        reward = (
            -self.lambda1 * (benign_exceedance_rate - self.alpha_rx) ** 2
            -self.lambda2 * (self.current_threshold / self.Tmax)
            -self.lambda3 * (delta_T ** 2)
        )

        self.current_index += 1
        self.episode_step += 1

        terminated = self.current_index >= len(self.rl_df)
        truncated = self.episode_step >= self.max_episode_steps

        if terminated or truncated:
            obs = np.zeros(len(self.feature_cols), dtype=np.float32)
        else:
            obs = self._get_obs()

        current_idx = min(self.current_index, len(self.rl_df) - 1)
        row = self.rl_df.iloc[current_idx]

        info = {
            "old_threshold": old_threshold,
            "delta_T": delta_T,
            "new_threshold": self.current_threshold,
            "benign_exceedance_rate": benign_exceedance_rate,
            "reward": reward,
            "index": self.current_index,
            "episode_step": self.episode_step,
            "envId": int(row["envId"]),
            "formationType": int(row["formationType"]),
            "timeStep": int(row["timeStep"]),
            "receiverId": int(row["receiverId"]),
        }

        return obs, reward, terminated, truncated, info


print("UAVThresholdEnv class is ready.")
```

```python
import numpy as np
import pandas as pd

from stable_baselines3 import SAC
from stable_baselines3.common.monitor import Monitor
from stable_baselines3.common.vec_env import DummyVecEnv, VecNormalize

# ============================================================
# SAC TRAINING SETUP - BALANCED DATASET VERSION
# ============================================================

# Create base environment
base_env = UAVThresholdEnv(
    rl_path="uav_rl_dataset_clean.csv",
    det_path="uav_detection_dataset-updated.csv",
    alpha_rx=0.05,
    lambda1=1.0,
    lambda2=0.05,
    lambda3=0.01,
    Tmin=1.0,
    Tmax=50.0,
    delta_limit=2.0,
    random_start=True,
    max_episode_steps=500
)

# Wrap environment for Stable-Baselines3
train_env = DummyVecEnv([
    lambda: Monitor(base_env)
])

# Normalize observations only.
# Do NOT normalize reward for this custom reward scale.
train_env = VecNormalize(
    train_env,
    norm_obs=True,
    norm_reward=False,
    clip_obs=10.0
)

# Create fresh SAC model
model = SAC(
    policy="MlpPolicy",
    env=train_env,
    learning_rate=3e-4,
    buffer_size=100000,
    learning_starts=1000,
    batch_size=256,
    tau=0.005,
    gamma=0.95,
    train_freq=1,
    gradient_steps=1,
    ent_coef="auto",
    verbose=1,
    seed=7,
)

print("Fresh SAC model created.")
```

```python
# ============================================================
# TRAIN SAC MODEL
# ============================================================

model.learn(
    total_timesteps=30000,
    progress_bar=True
)

# Save model and normalization stats
model.save("sac_uav_dynamic_threshold_model_balanced")
train_env.save("sac_uav_vecnormalize_balanced.pkl")

print("SAC balanced model saved.")
```

```python
import pandas as pd
import numpy as np

from stable_baselines3.common.monitor import Monitor
from stable_baselines3.common.vec_env import DummyVecEnv, VecNormalize

# ============================================================
# Correct evaluation using VecNormalize stats
# ============================================================

eval_base_env = UAVThresholdEnv(
    rl_path="uav_rl_dataset_clean.csv",
    det_path="uav_detection_dataset-updated.csv",
    alpha_rx=0.05,
    lambda1=1.0,
    lambda2=0.05,
    lambda3=0.01,
    Tmin=1.0,
    Tmax=50.0,
    delta_limit=2.0,
    random_start=False,
    max_episode_steps=28800
)

eval_vec_env = DummyVecEnv([
    lambda: Monitor(eval_base_env)
])

# Load the normalization statistics from training
eval_vec_env = VecNormalize.load(
    "sac_uav_vecnormalize_balanced.pkl",
    eval_vec_env
)

# Important for evaluation
eval_vec_env.training = False
eval_vec_env.norm_reward = False

obs = eval_vec_env.reset()

records = []
done = False
step_count = 0

while not done:
    action, _ = model.predict(obs, deterministic=True)

    obs, rewards, dones, infos = eval_vec_env.step(action)

    info = infos[0]
    reward = rewards[0]
    done = dones[0]

    records.append({
        "step": step_count,
        "envId": info["envId"],
        "formationType": info["formationType"],
        "timeStep": info["timeStep"],
        "receiverId": info["receiverId"],
        "old_threshold": info["old_threshold"],
        "delta_T": info["delta_T"],
        "new_threshold": info["new_threshold"],
        "benign_exceedance_rate": info["benign_exceedance_rate"],
        "reward": reward
    })

    step_count += 1

results_df = pd.DataFrame(records)

results_df.to_csv("sac_dynamic_threshold_results_balanced_normalized_eval.csv", index=False)

print("Saved:")
print("sac_dynamic_threshold_results_balanced_normalized_eval.csv")

print("\nResults shape:", results_df.shape)
print(results_df.head())
```
