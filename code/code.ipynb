## compute_exceedance

```python
import pandas as pd
import numpy as np

# =========================
# Load datasets
# =========================
rl_df = pd.read_csv("uav_rl_dataset_clean.csv")
det_df = pd.read_csv("uav_detection_dataset.csv")

print("RL shape:", rl_df.shape)
print("Detection shape:", det_df.shape)

# =========================
# Grouping columns
# =========================
group_cols = ["envId", "formationType", "timeStep", "receiverId"]

# =========================
# Keep only benign target rows for reward calculation
# =========================
benign_det = det_df[det_df["actual_target_malicious"] == 0].copy()

print("\nBenign detection rows:", benign_det.shape)

# =========================
# Compute exceedance rate using static theta from dataset
# Here theta = 10 in the generated RL dataset
# =========================

# Merge theta into detection rows based on receiver-time state
det_with_theta = benign_det.merge(
    rl_df[group_cols + ["theta"]],
    on=group_cols,
    how="inner"
)

print("\nDetection rows matched with RL theta:", det_with_theta.shape)

# Check whether delta_d exceeds current threshold
det_with_theta["exceeded_theta"] = (
    det_with_theta["delta_d"] > det_with_theta["theta"]
).astype(int)

# =========================
# Aggregate per receiver-time state
# =========================
exceedance_df = (
    det_with_theta
    .groupby(group_cols)
    .agg(
        benign_exceedance_rate=("exceeded_theta", "mean"),
        mean_delta_d=("delta_d", "mean"),
        max_delta_d=("delta_d", "max"),
        num_links=("delta_d", "count")
    )
    .reset_index()
)

print("\nExceedance dataframe shape:", exceedance_df.shape)

# =========================
# Merge exceedance info back into RL dataset
# =========================
rl_reward_df = rl_df.merge(
    exceedance_df,
    on=group_cols,
    how="left"
)

# Fill missing values if any
rl_reward_df["benign_exceedance_rate"] = rl_reward_df["benign_exceedance_rate"].fillna(0.0)
rl_reward_df["mean_delta_d"] = rl_reward_df["mean_delta_d"].fillna(0.0)
rl_reward_df["max_delta_d"] = rl_reward_df["max_delta_d"].fillna(0.0)
rl_reward_df["num_links"] = rl_reward_df["num_links"].fillna(0)

# =========================
# Print checks
# =========================
print("\nRL reward dataset shape:", rl_reward_df.shape)

print("\nColumns:")
print(rl_reward_df.columns.tolist())

print("\nBenign exceedance summary:")
print(rl_reward_df["benign_exceedance_rate"].describe())

print("\nMean delta_d summary:")
print(rl_reward_df["mean_delta_d"].describe())

print("\nNum links summary:")
print(rl_reward_df["num_links"].describe())

print("\nExceedance by environment:")
print(
    rl_reward_df
    .groupby("envId")["benign_exceedance_rate"]
    .describe()
)

print("\nExceedance by formation:")
print(
    rl_reward_df
    .groupby("formationType")["benign_exceedance_rate"]
    .describe()
)

# =========================
# Save output
# =========================
rl_reward_df.to_csv("uav_rl_reward_dataset.csv", index=False)

print("\nSaved:")
print("uav_rl_reward_dataset.csv")
```

## test_reward_function

```python
import pandas as pd
import numpy as np

# =========================
# Load datasets
# =========================
rl_df = pd.read_csv("uav_rl_dataset_clean.csv")
det_df = pd.read_csv("uav_detection_dataset.csv")

# =========================
# Reward parameters
# =========================
alpha_rx = 0.05     # desired benign exceedance rate
lambda1 = 1.0       # exceedance penalty
lambda2 = 0.05      # large-threshold penalty

# IMPORTANT:
# For manual fixed-threshold testing, keep lambda3 = 0.
# Threshold-change penalty is only meaningful during RL steps.
lambda3 = 0.0

Tmin = 1.0
Tmax = 50.0

thresholds_to_test = [2, 5, 8, 10, 12, 15, 20, 25, 30, 35, 40]

group_cols = ["envId", "formationType", "timeStep", "receiverId"]

# =========================
# Keep benign target rows only
# =========================
benign_det = det_df[det_df["actual_target_malicious"] == 0].copy()

results = []

# =========================
# Test multiple fixed thresholds
# =========================
for threshold in thresholds_to_test:

    temp = benign_det.copy()

    # Check which benign links exceed this threshold
    temp["exceeded"] = (temp["delta_d"] > threshold).astype(int)

    # Compute exceedance rate per receiver-time state
    exceedance_df = (
        temp
        .groupby(group_cols)
        .agg(
            benign_exceedance_rate=("exceeded", "mean"),
            mean_delta_d=("delta_d", "mean"),
            max_delta_d=("delta_d", "max"),
            num_links=("delta_d", "count")
        )
        .reset_index()
    )

    # Only evaluate states that exist in the RL dataset
    merged = rl_df[group_cols].merge(
        exceedance_df,
        on=group_cols,
        how="inner"
    )

    # For manual testing, no threshold-change penalty
    delta_T = 0.0

    merged["reward"] = (
        -lambda1 * (merged["benign_exceedance_rate"] - alpha_rx) ** 2
        -lambda2 * (threshold / Tmax)
        -lambda3 * (delta_T ** 2)
    )

    summary = (
        merged
        .groupby("envId")
        .agg(
            threshold=("benign_exceedance_rate", lambda x: threshold),
            mean_exceedance=("benign_exceedance_rate", "mean"),
            std_exceedance=("benign_exceedance_rate", "std"),
            mean_reward=("reward", "mean"),
            rows=("reward", "count")
        )
        .reset_index()
    )

    results.append(summary)

# =========================
# Combine all threshold results
# =========================
reward_results = pd.concat(results, ignore_index=True)

print("===== REWARD RESULTS BY ENVIRONMENT =====")
print(reward_results)

# =========================
# Best threshold per environment
# =========================
best_by_env = (
    reward_results
    .sort_values(["envId", "mean_reward"], ascending=[True, False])
    .groupby("envId")
    .head(1)
)

print("\n===== BEST THRESHOLD PER ENVIRONMENT =====")
print(best_by_env)

# =========================
# Optional: best threshold per environment + formation
# =========================
formation_results = []

for threshold in thresholds_to_test:

    temp = benign_det.copy()
    temp["exceeded"] = (temp["delta_d"] > threshold).astype(int)

    exceedance_df = (
        temp
        .groupby(group_cols)
        .agg(
            benign_exceedance_rate=("exceeded", "mean"),
            mean_delta_d=("delta_d", "mean"),
            max_delta_d=("delta_d", "max"),
            num_links=("delta_d", "count")
        )
        .reset_index()
    )

    merged = rl_df[group_cols].merge(
        exceedance_df,
        on=group_cols,
        how="inner"
    )

    merged["reward"] = (
        -lambda1 * (merged["benign_exceedance_rate"] - alpha_rx) ** 2
        -lambda2 * (threshold / Tmax)
    )

    form_summary = (
        merged
        .groupby(["envId", "formationType"])
        .agg(
            threshold=("benign_exceedance_rate", lambda x: threshold),
            mean_exceedance=("benign_exceedance_rate", "mean"),
            mean_reward=("reward", "mean"),
            rows=("reward", "count")
        )
        .reset_index()
    )

    formation_results.append(form_summary)

formation_reward_results = pd.concat(formation_results, ignore_index=True)

best_by_env_formation = (
    formation_reward_results
    .sort_values(["envId", "formationType", "mean_reward"], ascending=[True, True, False])
    .groupby(["envId", "formationType"])
    .head(1)
)

print("\n===== BEST THRESHOLD PER ENVIRONMENT + FORMATION =====")
print(best_by_env_formation)

# =========================
# Save outputs
# =========================
reward_results.to_csv("reward_threshold_test_results.csv", index=False)
best_by_env.to_csv("best_threshold_by_environment.csv", index=False)
formation_reward_results.to_csv("reward_threshold_test_by_env_formation.csv", index=False)
best_by_env_formation.to_csv("best_threshold_by_env_formation.csv", index=False)

print("\nSaved:")
print("reward_threshold_test_results.csv")
print("best_threshold_by_environment.csv")
print("reward_threshold_test_by_env_formation.csv")
print("best_threshold_by_env_formation.csv")
```

## environment class code

```python
import numpy as np
import pandas as pd
import gymnasium as gym
from gymnasium import spaces


class UAVThresholdEnv(gym.Env):
    """
    Custom environment for UAV dynamic Phase-1 threshold learning.

    Observation:
        Receiver-level RL features.

    Action:
        Continuous threshold adjustment delta_T.

    Reward:
        Penalizes difference between benign exceedance rate and target alpha_rx,
        penalizes large threshold,
        penalizes abrupt threshold changes.
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        rl_path="uav_rl_dataset_clean.csv",
        det_path="uav_detection_dataset.csv",
        alpha_rx=0.05,
        lambda1=1.0,
        lambda2=0.05,
        lambda3=0.01,
        Tmin=1.0,
        Tmax=50.0,
        delta_limit=2.0,
    ):
        super().__init__()

        self.rl_df = pd.read_csv(rl_path)
        self.det_df = pd.read_csv(det_path)

        self.alpha_rx = alpha_rx
        self.lambda1 = lambda1
        self.lambda2 = lambda2
        self.lambda3 = lambda3

        self.Tmin = Tmin
        self.Tmax = Tmax
        self.delta_limit = delta_limit

        self.group_cols = ["envId", "formationType", "timeStep", "receiverId"]

        self.feature_cols = [
            "sigma_rssi_bar",
            "sigma_gps_bar",
            "snr_bar",
            "plr_bar",
            "relative_speed_bar",
            "relative_height_bar",
            "trusted_count",
            "trust_bar",
            "theta",
        ]

        # Keep only benign detection rows for reward
        self.benign_det = self.det_df[
            self.det_df["actual_target_malicious"] == 0
        ].copy()

        # Pre-group delta_d values for fast lookup
        self.delta_lookup = {
            key: group["delta_d"].values.astype(np.float32)
            for key, group in self.benign_det.groupby(self.group_cols)
        }

        # Observation space
        self.observation_space = spaces.Box(
            low=-np.inf,
            high=np.inf,
            shape=(len(self.feature_cols),),
            dtype=np.float32,
        )

        # Action = delta threshold
        self.action_space = spaces.Box(
            low=np.array([-self.delta_limit], dtype=np.float32),
            high=np.array([self.delta_limit], dtype=np.float32),
            dtype=np.float32,
        )

        self.current_index = 0
        self.current_threshold = None

    def _get_obs(self):
        row = self.rl_df.iloc[self.current_index]
        obs = row[self.feature_cols].values.astype(np.float32)

        # Replace theta feature with current dynamic threshold
        obs[-1] = self.current_threshold

        return obs

    def _get_key(self):
        row = self.rl_df.iloc[self.current_index]
        return tuple(row[col] for col in self.group_cols)

    def _compute_exceedance_rate(self, threshold):
        key = self._get_key()

        delta_values = self.delta_lookup.get(key, None)

        if delta_values is None or len(delta_values) == 0:
            return 0.0

        return float(np.mean(delta_values > threshold))

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)

        self.current_index = 0

        # Start from theta in dataset
        self.current_threshold = float(self.rl_df.iloc[self.current_index]["theta"])

        obs = self._get_obs()

        info = {
            "index": self.current_index,
            "threshold": self.current_threshold,
        }

        return obs, info

    def step(self, action):
        # Convert action to scalar delta_T
        delta_T = float(np.clip(action[0], -self.delta_limit, self.delta_limit))

        old_threshold = self.current_threshold

        # Update threshold
        self.current_threshold = float(
            np.clip(
                self.current_threshold + delta_T,
                self.Tmin,
                self.Tmax
            )
        )

        # Compute benign exceedance rate using updated threshold
        benign_exceedance_rate = self._compute_exceedance_rate(self.current_threshold)

        # Reward function
        reward = (
            -self.lambda1 * (benign_exceedance_rate - self.alpha_rx) ** 2
            -self.lambda2 * (self.current_threshold / self.Tmax)
            -self.lambda3 * (delta_T ** 2)
        )

        # Move to next row
        self.current_index += 1

        terminated = self.current_index >= len(self.rl_df)
        truncated = False

        if terminated:
            # return zero obs at end
            obs = np.zeros(len(self.feature_cols), dtype=np.float32)
        else:
            obs = self._get_obs()

        info = {
            "old_threshold": old_threshold,
            "delta_T": delta_T,
            "new_threshold": self.current_threshold,
            "benign_exceedance_rate": benign_exceedance_rate,
            "reward": reward,
            "index": self.current_index,
        }

        return obs, reward, terminated, truncated, info


# =========================================================
```

## uav_threshold_env_test

```python
import numpy as np
import pandas as pd
import gymnasium as gym
from gymnasium import spaces


class UAVThresholdEnv(gym.Env):
    """
    Custom Gymnasium environment for UAV dynamic threshold learning.

    State:
        Receiver-level RL features from uav_rl_dataset_clean.csv

    Action:
        Continuous threshold adjustment delta_T

    Reward:
        Keeps benign exceedance rate close to alpha_rx,
        while penalizing too-large thresholds and abrupt threshold changes.
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        rl_path="uav_rl_dataset_clean.csv",
        det_path="uav_detection_dataset.csv",
        alpha_rx=0.05,
        lambda1=1.0,
        lambda2=0.05,
        lambda3=0.01,
        Tmin=1.0,
        Tmax=50.0,
        delta_limit=2.0,
    ):
        super().__init__()

        # =========================
        # Load datasets
        # =========================
        self.rl_df = pd.read_csv(rl_path)
        self.det_df = pd.read_csv(det_path)

        # =========================
        # Reward parameters
        # =========================
        self.alpha_rx = alpha_rx
        self.lambda1 = lambda1
        self.lambda2 = lambda2
        self.lambda3 = lambda3

        # =========================
        # Threshold parameters
        # =========================
        self.Tmin = Tmin
        self.Tmax = Tmax
        self.delta_limit = delta_limit

        # =========================
        # Columns
        # =========================
        self.group_cols = ["envId", "formationType", "timeStep", "receiverId"]

        self.feature_cols = [
            "sigma_rssi_bar",
            "sigma_gps_bar",
            "snr_bar",
            "plr_bar",
            "relative_speed_bar",
            "relative_height_bar",
            "trusted_count",
            "trust_bar",
            "theta",
        ]

        # =========================
        # Keep benign rows only
        # =========================
        self.benign_det = self.det_df[
            self.det_df["actual_target_malicious"] == 0
        ].copy()

        # =========================
        # Fast lookup:
        # key = (envId, formationType, timeStep, receiverId)
        # value = all delta_d values for that receiver-time state
        # =========================
        self.delta_lookup = {
            key: group["delta_d"].values.astype(np.float32)
            for key, group in self.benign_det.groupby(self.group_cols)
        }

        # =========================
        # Observation space
        # 9 features
        # =========================
        self.observation_space = spaces.Box(
            low=-np.inf,
            high=np.inf,
            shape=(len(self.feature_cols),),
            dtype=np.float32,
        )

        # =========================
        # Action space
        # SAC outputs delta_T in [-delta_limit, +delta_limit]
        # =========================
        self.action_space = spaces.Box(
            low=np.array([-self.delta_limit], dtype=np.float32),
            high=np.array([self.delta_limit], dtype=np.float32),
            dtype=np.float32,
        )

        # =========================
        # Internal state
        # =========================
        self.current_index = 0
        self.current_threshold = None

    def _get_obs(self):
        """
        Return current observation.
        The last state feature is theta/current threshold.
        We replace it with the current dynamic threshold.
        """
        row = self.rl_df.iloc[self.current_index]
        obs = row[self.feature_cols].values.astype(np.float32)

        # Replace static theta with dynamic threshold
        obs[-1] = self.current_threshold

        return obs

    def _get_key(self):
        """
        Return key for matching RL row to detection rows.
        """
        row = self.rl_df.iloc[self.current_index]
        return tuple(row[col] for col in self.group_cols)

    def _compute_exceedance_rate(self, threshold):
        """
        Compute benign exceedance rate for current receiver-time state.

        benign_exceedance_rate = fraction of benign links where delta_d > threshold
        """
        key = self._get_key()

        delta_values = self.delta_lookup.get(key, None)

        if delta_values is None or len(delta_values) == 0:
            return 0.0

        return float(np.mean(delta_values > threshold))

    def reset(self, seed=None, options=None):
        """
        Start episode from first row.
        """
        super().reset(seed=seed)

        self.current_index = 0

        # Start threshold from theta column in dataset
        self.current_threshold = float(self.rl_df.iloc[self.current_index]["theta"])

        obs = self._get_obs()

        info = {
            "index": self.current_index,
            "threshold": self.current_threshold,
        }

        return obs, info

    def step(self, action):
        """
        Apply threshold adjustment and compute reward.
        """

        # =========================
        # Convert action to scalar
        # =========================
        delta_T = float(np.clip(action[0], -self.delta_limit, self.delta_limit))

        old_threshold = self.current_threshold

        # =========================
        # Update threshold
        # =========================
        self.current_threshold = float(
            np.clip(
                self.current_threshold + delta_T,
                self.Tmin,
                self.Tmax,
            )
        )

        # =========================
        # Compute benign exceedance
        # =========================
        benign_exceedance_rate = self._compute_exceedance_rate(
            self.current_threshold
        )

        # =========================
        # Reward
        # =========================
        reward = (
            -self.lambda1 * (benign_exceedance_rate - self.alpha_rx) ** 2
            -self.lambda2 * (self.current_threshold / self.Tmax)
            -self.lambda3 * (delta_T ** 2)
        )

        # =========================
        # Move to next row
        # =========================
        self.current_index += 1

        terminated = self.current_index >= len(self.rl_df)
        truncated = False

        if terminated:
            obs = np.zeros(len(self.feature_cols), dtype=np.float32)
        else:
            obs = self._get_obs()

        info = {
            "old_threshold": old_threshold,
            "delta_T": delta_T,
            "new_threshold": self.current_threshold,
            "benign_exceedance_rate": benign_exceedance_rate,
            "reward": reward,
            "index": self.current_index,
        }

        return obs, reward, terminated, truncated, info


# ============================================================
# Manual test only
# ============================================================
if __name__ == "__main__":

    env = UAVThresholdEnv()

    obs, info = env.reset()

    print("===== ENV RESET =====")
    print("Initial observation:")
    print(obs)
    print("Info:")
    print(info)

    print("\n===== MANUAL STEPS =====")

    # Test different actions manually
    test_actions = [
        np.array([0.0], dtype=np.float32),    # keep threshold same
        np.array([1.0], dtype=np.float32),    # increase threshold
        np.array([-1.0], dtype=np.float32),   # decrease threshold
        np.array([2.0], dtype=np.float32),    # max increase
        np.array([-2.0], dtype=np.float32),   # max decrease
    ]

    for action in test_actions:
        obs, reward, terminated, truncated, info = env.step(action)

        print("\nAction:", action)
        print("Reward:", reward)
        print("Info:", info)
        print("Next obs:", obs)

        if terminated:
            print("Episode finished.")
            break
```

```python
import numpy as np
import pandas as pd
import gymnasium as gym
from gymnasium import spaces


class UAVThresholdEnv(gym.Env):
    """
    Custom Gymnasium environment for UAV dynamic Phase-1 threshold learning.

    Observation:
        Receiver-level RL features.

    Action:
        Continuous threshold adjustment delta_T.

    Reward:
        Penalizes difference between benign exceedance rate and target alpha_rx,
        penalizes large threshold,
        penalizes abrupt threshold changes.

    Episode design:
        reset() can start from a random row.
        step() then moves sequentially for max_episode_steps.
    """

    metadata = {"render_modes": []}

    def __init__(
        self,
        rl_path="uav_rl_dataset_clean.csv",
        det_path="uav_detection_dataset.csv",
        alpha_rx=0.05,
        lambda1=1.0,
        lambda2=0.05,
        lambda3=0.01,
        Tmin=1.0,
        Tmax=50.0,
        delta_limit=2.0,
        random_start=True,
        max_episode_steps=500,
    ):
        super().__init__()

        # =========================
        # Load datasets
        # =========================
        self.rl_df = pd.read_csv(rl_path)
        self.det_df = pd.read_csv(det_path)

        # =========================
        # Reward parameters
        # =========================
        self.alpha_rx = alpha_rx
        self.lambda1 = lambda1
        self.lambda2 = lambda2
        self.lambda3 = lambda3

        # =========================
        # Threshold parameters
        # =========================
        self.Tmin = Tmin
        self.Tmax = Tmax
        self.delta_limit = delta_limit

        # =========================
        # Episode control
        # =========================
        self.random_start = random_start
        self.max_episode_steps = max_episode_steps
        self.episode_step = 0

        # =========================
        # Columns
        # =========================
        self.group_cols = ["envId", "formationType", "timeStep", "receiverId"]

        self.feature_cols = [
            "sigma_rssi_bar",
            "sigma_gps_bar",
            "snr_bar",
            "plr_bar",
            "relative_speed_bar",
            "relative_height_bar",
            "trusted_count",
            "trust_bar",
            "theta",
        ]

        # =========================
        # Keep benign rows only for reward
        # =========================
        self.benign_det = self.det_df[
            self.det_df["actual_target_malicious"] == 0
        ].copy()

        # =========================
        # Fast lookup:
        # key = (envId, formationType, timeStep, receiverId)
        # value = all delta_d values for that receiver-time state
        # =========================
        self.delta_lookup = {
            key: group["delta_d"].values.astype(np.float32)
            for key, group in self.benign_det.groupby(self.group_cols)
        }

        # =========================
        # Observation space
        # =========================
        self.observation_space = spaces.Box(
            low=-np.inf,
            high=np.inf,
            shape=(len(self.feature_cols),),
            dtype=np.float32,
        )

        # =========================
        # Action space
        # =========================
        self.action_space = spaces.Box(
            low=np.array([-self.delta_limit], dtype=np.float32),
            high=np.array([self.delta_limit], dtype=np.float32),
            dtype=np.float32,
        )

        # =========================
        # Internal state
        # =========================
        self.current_index = 0
        self.start_index = 0
        self.current_threshold = None

    def _get_obs(self):
        """
        Return current observation.
        The last feature is current dynamic threshold.
        """
        row = self.rl_df.iloc[self.current_index]
        obs = row[self.feature_cols].values.astype(np.float32)

        # Replace static theta with dynamic threshold
        obs[-1] = self.current_threshold

        return obs

    def _get_key(self):
        """
        Return key for matching RL row to detection rows.
        """
        row = self.rl_df.iloc[self.current_index]
        return tuple(row[col] for col in self.group_cols)

    def _compute_exceedance_rate(self, threshold):
        """
        benign_exceedance_rate = fraction of benign links where delta_d > threshold
        """
        key = self._get_key()
        delta_values = self.delta_lookup.get(key, None)

        if delta_values is None or len(delta_values) == 0:
            return 0.0

        return float(np.mean(delta_values > threshold))

    def reset(self, seed=None, options=None):
        """
        Start a new episode.
        If random_start=True, start from a random row but still move sequentially after that.
        """
        super().reset(seed=seed)

        self.episode_step = 0

        if self.random_start:
            max_start = max(0, len(self.rl_df) - self.max_episode_steps - 1)
            self.start_index = int(self.np_random.integers(0, max_start + 1))
        else:
            self.start_index = 0

        self.current_index = self.start_index

        # Start threshold from theta in dataset
        self.current_threshold = float(self.rl_df.iloc[self.current_index]["theta"])

        obs = self._get_obs()

        row = self.rl_df.iloc[self.current_index]

        info = {
            "start_index": self.start_index,
            "index": self.current_index,
            "threshold": self.current_threshold,
            "envId": int(row["envId"]),
            "formationType": int(row["formationType"]),
            "timeStep": int(row["timeStep"]),
            "receiverId": int(row["receiverId"]),
        }

        return obs, info

    def step(self, action):
        """
        Apply threshold adjustment and compute reward.
        """
        # Convert action to scalar delta_T
        delta_T = float(np.clip(action[0], -self.delta_limit, self.delta_limit))

        old_threshold = self.current_threshold

        # Update threshold
        self.current_threshold = float(
            np.clip(
                self.current_threshold + delta_T,
                self.Tmin,
                self.Tmax,
            )
        )

        # Compute benign exceedance rate using updated threshold
        benign_exceedance_rate = self._compute_exceedance_rate(
            self.current_threshold
        )

        # Reward
        reward = (
            -self.lambda1 * (benign_exceedance_rate - self.alpha_rx) ** 2
            -self.lambda2 * (self.current_threshold / self.Tmax)
            -self.lambda3 * (delta_T ** 2)
        )

        # Move forward sequentially
        self.current_index += 1
        self.episode_step += 1

        terminated = self.current_index >= len(self.rl_df)
        truncated = self.episode_step >= self.max_episode_steps

        if terminated or truncated:
            obs = np.zeros(len(self.feature_cols), dtype=np.float32)
        else:
            obs = self._get_obs()

        current_idx = min(self.current_index, len(self.rl_df) - 1)
        row = self.rl_df.iloc[current_idx]

        info = {
            "old_threshold": old_threshold,
            "delta_T": delta_T,
            "new_threshold": self.current_threshold,
            "benign_exceedance_rate": benign_exceedance_rate,
            "reward": reward,
            "index": self.current_index,
            "episode_step": self.episode_step,
            "envId": int(row["envId"]),
            "formationType": int(row["formationType"]),
            "timeStep": int(row["timeStep"]),
            "receiverId": int(row["receiverId"]),
        }

        return obs, reward, terminated, truncated, info
```
