```python
import pandas as pd
import numpy as np

results_df = pd.read_csv("sac_dynamic_threshold_results.csv")

print("Results shape:", results_df.shape)
print(results_df.head())

print("\n===== SAC THRESHOLD BY ENVIRONMENT =====")
print(
    results_df
    .groupby("envId")["new_threshold"]
    .describe()
)

print("\n===== SAC EXCEEDANCE BY ENVIRONMENT =====")
print(
    results_df
    .groupby("envId")["benign_exceedance_rate"]
    .describe()
)

print("\n===== SAC REWARD BY ENVIRONMENT =====")
print(
    results_df
    .groupby("envId")["reward"]
    .describe()
)
```
